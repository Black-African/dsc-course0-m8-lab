## Data Cleaning

### Filtering aircrafts and events

We want to filter the dataset to include aircraft that the client is interested in an analysis of:
- inspect relevant columns
- figure out any reasonable imputations
- filter the dataset

```python
# Inspecting the columns relevant to filtering aircraft and events

data_columns = [
    "Event.Date",
    "Investigation.Type",
    "Amateur.Built",
    "Aircraft.Category",
    "Make",
    "Model"
]

df[data_columns].head()

print('-'*40)

# Checking data types and missing values in the relevant columns

df[data_columns].info()

df[data_columns].isna().sum()

# Examining values to identify inconsistencies and possible imputations

for col in data_columns:
    print(f"\n {col} value counts:")
    print(df[col].value_counts(dropna=False).head())

print('-'*40)
# Standardizing aircraft category values

df["Aircraft.Category"] = df["Aircraft.Category"].replace({
    "UNK": np.nan,
    "Unknown": np.nan,
    "WSFT": "Weight-Shift",
    "ULTR": "Ultralight"
})

df["Aircraft.Category"].value_counts(dropna=False)

print('-'*40)
# Filling missing categories using the mode for each model

model_category_mode = (
    df.dropna(subset=["Model", "Aircraft.Category"])
      .groupby("Model")["Aircraft.Category"]
      .agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan)
)

missing_category = df["Aircraft.Category"].isna()

df.loc[missing_category, "Aircraft.Category"] = (
    df.loc[missing_category, "Model"].map(model_category_mode)
)

df["Aircraft.Category"].isna().sum()
print('-'*40)
# Converting Event.Date to datetime

df["Event.Date"] = pd.to_datetime(
    df["Event.Date"],
    errors="coerce"
)

```

```python
# Filtering records according to the client's criteria

df = df[
    (df["Event.Date"] >= "1983-01-01") &
    (df["Investigation.Type"] == "Accident") &
    (df["Amateur.Built"] == "No") &
    (df["Aircraft.Category"] == "Airplane")
].copy()

df.head()
print('-'*40)

print("Rows and columns after filtering:", df.shape)

print("\nInvestigation types:")
print(df["Investigation.Type"].value_counts(dropna=False))

print("\nAmateur-built status:")
print(df["Amateur.Built"].value_counts(dropna=False))

print("\nAircraft categories:")
print(df["Aircraft.Category"].value_counts(dropna=False))

print("\nEarliest event date:", df["Event.Date"].min())
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python

injury_columns = [
    "Injury.Severity",
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

# Displaying a few records to understand the data.
df[injury_columns].head(10)
print('-'*40)
# Checking which injury columns have missing values and their data types.
print("Missing values in each injury column:")
print(df[injury_columns].isna().sum())

print("\nData types:")
print(df[injury_columns].dtypes)
print('-'*40)
# Create a temporary estimate of fatal injuries using the severity label.
severity_fatal = (
    df["Injury.Severity"]
    .astype("string")
    .str.contains("Fatal", case=False, na=False)
)

reported_fatal = df["Total.Fatal.Injuries"]

# Comparing records where both sources provide enough information.
check_records = reported_fatal.notna() & df["Injury.Severity"].notna()

if check_records.sum() > 0:
    agreement = (
        (reported_fatal[check_records] > 0)
        == severity_fatal[check_records]
    ).mean()

    print(f"Agreement between severity and fatal counts: {agreement:.2%}")
else:
    print("There are not enough complete records to compare.")
print('-'*40)
# Only use the severity-based estimate if the agreement exceeds 99%.
if check_records.sum() > 0 and agreement > 0.99:

    # Estimating one or more fatalities when severity is marked as fatal.
    # For non-fatal severity labels, the estimate is zero.
    estimated_fatal = severity_fatal.astype(int)

    # Filling only missing fatal counts; preserve reported values.
    missing_fatal = df["Total.Fatal.Injuries"].isna()

    df.loc[missing_fatal, "Total.Fatal.Injuries"] = (
        estimated_fatal[missing_fatal]
    )

    print("Missing fatal injury counts have been filled.")
else:
    print(
        "Agreement did not exceed 99%. "
        "Review the severity labels before imputing fatal counts."
    )

print('-'*40)
# These four columns describe the different injury outcomes.
count_columns = [
    "Total.Fatal.Injuries",
    "Total.Serious.Injuries",
    "Total.Minor.Injuries",
    "Total.Uninjured"
]

# Counting how many records have no injury information at all.
all_counts_missing = df[count_columns].isna().all(axis=1)

print("Records with all injury counts missing:", all_counts_missing.sum())

# Removing records where none of the four counts is available.
df = df.loc[~all_counts_missing].copy()
print('-'*40)
# Assumption: For records with at least one reported count,
# a blank in another injury-count column is treated as zero.

df[count_columns] = df[count_columns].fillna(0)

# Injury counts should be whole numbers, not decimals.
df[count_columns] = df[count_columns].astype(int)

# Confirming that no missing injury counts remain.
print(df[count_columns].isna().sum())
print(df[count_columns].dtypes)
print('-'*40)


df["Total.Passengers"] = df[count_columns].sum(axis=1)

# Inspecting the new column.
df[["Total.Passengers"] + count_columns].head(5)
print('-'*40)
# A meaningful injury fraction requires a positive total.
df = df[df["Total.Passengers"] > 0].copy()

print("Dataset shape after injury-count cleaning:", df.shape)
print('-'*40)
# Derived column: Fatal.Serious.Fraction
# A value close to 1 means a high proportion of people suffered
# fatal or serious injuries. A value close to 0 means a lower proportion.

df["Fatal.Serious.Fraction"] = (
    df["Total.Fatal.Injuries"] + df["Total.Serious.Injuries"]
) / df["Total.Passengers"]

# Displaying the results for a quick review.
df[
    [
        "Total.Fatal.Injuries",
        "Total.Serious.Injuries",
        "Total.Passengers",
        "Fatal.Serious.Fraction"
    ]
].head(5)
print('-'*40)
# Checking that the fraction is between 0 and 1.
print(
    "Fraction within the expected range:",
    df["Fatal.Serious.Fraction"].between(0, 1).all()
)

# Checking for missing values in the new columns.
print("\nMissing values in derived columns:")
print(
    df[["Total.Passengers", "Fatal.Serious.Fraction"]]
    .isna()
    .sum()
)

# Reviewing summarizwd statistics for the new metric.
df[["Total.Passengers", "Fatal.Serious.Fraction"]].describe()
```

### Cleaning tasks for the Make column

1. **Inspect the column:** Check the number of missing values, unique manufacturers, and most frequent make names.
2. **Handle missing values:** Remove records with missing or blank manufacturer names because they cannot be reliably assigned to a manufacturer.
3. **Standardize text:** Remove leading and trailing spaces and convert manufacturer names to uppercase for consistency.
4. **Correct inconsistent labels:** Apply targeted replacements for known aliases and spelling variations so that the same manufacturer is not counted multiple times.
5. **Inspect manufacturer counts:** Check how frequently each manufacturer appears after cleaning.
6. **Filter uncommon manufacturers:** Keep manufacturers with at least 50 records to focus the analysis on makes with a reasonable amount of data.

**Assumption:** The threshold of 50 records is a practical choice for this analysis, not a claim that less frequent manufacturers are invalid. Filtering will reduce the scope of the analysis to more frequently recorded manufacturers.

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python

# Cleaning the Model column by removing unnecessary spaces.
df["Model"] = (
    df["Model"]
    .astype("string")
    .str.strip()
)

# Treating empty model names as missing values.
df["Model"] = df["Model"].replace("", pd.NA)

# Removing records without a model name.
df = df.dropna(subset=["Model"]).copy()

# Checking the results.
print("Missing model values:", df["Model"].isna().sum())
print("Remaining records:", len(df))


# Counting how frequently each model appears.
model_counts = df["Model"].value_counts()

print("Number of unique models:", df["Model"].nunique())

print("\nTop 20 most common models:")
print(model_counts.head(20))


# Counting the records for each manufacturer and model combination.
make_model_counts = (
    df.groupby(["Make", "Model"], dropna=False)
      .size()
      .reset_index(name="Record.Count")
      .sort_values("Record.Count", ascending=False)
)

print("Top 20 make/model combinations:")
display(make_model_counts.head(20))


# Counting how many different manufacturers use each model label.
makes_per_model = df.groupby("Model")["Make"].nunique()

# Finding model labels associated with more than one manufacturer.
shared_models = makes_per_model[makes_per_model > 1]

print("Model labels used by multiple manufacturers:", len(shared_models))

print("\nExamples of shared model labels:")
display(shared_models.sort_values(ascending=False).head(20))


# Standardizing manufacturer and model labels.
df["Make"] = df["Make"].astype("string").str.strip().str.upper()
df["Model"] = df["Model"].astype("string").str.strip().str.upper()

# Removing records without a manufacturer or model.
df = df.dropna(subset=["Make", "Model"]).copy()

# Replacing spaces within model names with hyphens.
df["Model"] = df["Model"].str.replace(r"\s+", "-", regex=True)

# Creating a combined aircraft-type identifier.
df["Make.Model"] = df["Make"] + "." + df["Model"]

# Inspecting the new column.
display(df[["Make", "Model", "Make.Model"]].head(10))

print("Unique aircraft types:", df["Make.Model"].nunique())



# Confirming that each identifier maps to one make and one model.
identifier_check = (
    df.groupby("Make.Model")[["Make", "Model"]]
      .nunique()
)

print(
    "Identifiers associated with multiple makes:",
    (identifier_check["Make"] > 1).sum()
)

print(
    "Identifiers associated with multiple models:",
    (identifier_check["Model"] > 1).sum()
)
```
